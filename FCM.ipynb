Implementation of Fuzzy C-Means and K-Means.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv('/content/drive/MyDrive/A2-Q1.csv')
X0=data['X0'].values
X1=data['X1'].values
X=np.column_stack((X0,X1))
```

```python
print(np.shape(X))
```

```python
plt.figure
plt.scatter(X[:,0],X[:,1])
plt.title("Data Points")
plt.xlabel('X0')
plt.ylabel('X1')
plt.show()
```

```python

def centroid(X,m,U):
  # Finding the centroid of the clusters
  Um=(U**m).T
  centroids=np.zeros((np.shape(U)[1],np.shape(X)[1]))

  for j in range(np.shape(U)[1]): # Finding centroid of every class
      centroids[j,:]=(Um[j,:]@X)/(np.sum(Um[j,:]))
  return centroids




```

```python

def membership(X,m,centroids):
  # Updating the membership matrix using the calculated centroids
  ns=np.shape(X)[0]
  C=np.shape(centroids)[0]
  U=np.zeros((ns,C))
  for i in range(ns):
    for j in range (C):
      dist_xiCj=np.linalg.norm(X[i,:]-centroids[j,:]) # Calculate the distance between the data point i and centroid of class j

      sum=0.0
      for k in range(C):
        dist_xiCk=np.linalg.norm(X[i,:]-centroids[k,:]) # Calculate the distance between the data point i and centroid of class k
        sum=sum+(dist_xiCj/dist_xiCk)**(2/(m-1))
      U[i,j]=1/sum
  return U


```

```python
C=3
m=1.618
np.random.seed(42)
U = np.random.dirichlet(np.ones(C), np.shape(X)[0]) # Creates random values such that sum of rows is always equal to one
print(np.shape(U))

for i in range(200):
  U_=U.copy()
  centroids=centroid(X,m,U)
  U=membership(X,m,centroids)
  if np.linalg.norm(U-U_)<0.00001:
    print(i)
    break
print(centroids)
plt.figure
plt.scatter(X[:,0],X[:,1],c=np.argmax(U,axis=1))
plt.scatter(centroids[:, 0], centroids[:, 1],
            marker='o', s=50, c='red', edgecolors='black', label='Centroids')


plt.title("Fuzzy C Means")
plt.xlabel('X0')
plt.ylabel('X1')
plt.show()
```

```python



sns.heatmap(U,
            cmap=None,
            xticklabels=[f'Cluster {i}' for i in range(1,C+1)],
            cbar_kws={'label': 'Membership Probability ($u_{ij}$)'})

plt.title("Membership Matrix Heatmap")
plt.xlabel("Cluster Index")
plt.ylabel("Point Index (1 to 1000)")
plt.show()
```

```python
def kmeans(X,k):
  ns=np.shape(X)[0]
  distances=np.zeros((ns,k))
  np.random.seed(0)
  centroidsk=X[np.random.choice(ns,k,replace=False),:] # Randomly taking points as centroid of cluster

  for iter in range(200):
    old_centroids=centroidsk.copy()
    for i in range(k):
      for j in range(ns):
        distance=np.linalg.norm(X[j]-centroidsk[i])
        distances[j,i]=distance
    labels=np.argmin(distances,axis=1) # Labels for each clusters
    new_center=np.zeros((k,np.shape(X)[1]))
    for i in range(k):
      cluster_points=X[labels==i]
      if np.shape(cluster_points)[0]>0:
        new_center[i]=np.mean(cluster_points,axis=0) # New centroids are being updated
      else:
        new_center[i]=old_centroids[i] # if there is no points in the cluster then old centroid is assigned
    centroidsk=new_center
    if np.linalg.norm(centroidsk-old_centroids)<0.00001:
      print(iter)
      break
  return centroidsk,labels
```

```python
centroidsk,labels=kmeans(X,3)
plt.figure
plt.scatter(X[:,0],X[:,1],c=labels)
plt.scatter(centroidsk[:, 0], centroidsk[:, 1],
            marker='o', s=50, c='red', edgecolors='black', label='Centroids')


plt.title("K Means Clustering")
plt.xlabel('X0')
plt.ylabel('X1')
plt.show()
print(centroidsk)

```
