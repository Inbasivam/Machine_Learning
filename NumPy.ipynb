The main object in numpy is the ndarray(N dimensional array) and it is multidimensional and homogenous array with predetermined number of items. All the items present in the array are of same type and size. The data type is specified by another object called dtype(data-type). The number of dimensions and items are defined by its shape, a tuple of N-dimensional items specifies the size for each dimension. The dimensions are defined as axes and the number of axes as rank.
Also the size of the array is fixed from the time of creation and doesn't change.

Arthimetic operations in arrays. The * means elementwise operations in array and adding or multiplying a scalar simply means elementwise operations.

```python
a1=np.array([2,5,8,10])
b1=a1+4
c1=a1*4
d1=np.array([1,2,3,4])
a2=a1*d1 # you can also use multiply function here as well as add, divide, subtract functions depending on the use case
print(b1,c1,a2)
```

To get the matrix multiplication use dot function. You can also use @ operator

```python
A=np.linspace(0,2,3) #row vector
B=np.ones((3,1))
C=np.dot(A,B)
D=A.dot(B) # alternate
E=A@B # alternate
print(C,D,E)
```

The same increment and decrement operators works here such as +=, -=, *=.

Universal Functions(ufunc) which operates element-by-element operations on array such log, sin, sqrt, etc. While the aggregate functions perform an operation on set of values an array and provide a single output examples include sum, min, max, mean and standard deviation.

Indexing uses square brackets to retrieve the elements for 2 dimensional arrays indexing is similar to index in matrix elements.

```python
E=np.arange(1,10).reshape(3,3)
print(E)
print(E[0,2])
```

Slicing for 1d array is similar to slicing in lists. We will see some examples for 2d arrays.

```python
print(E[0,:]) #returns the 1st row
print(E[:,0]) #returns the 1st column
print(E[1:3,1:3])
print(E[[0,2],0:2]) #you can also select different elements from different rows you can see that using slicing we get a higher rank array than using indexing.
```

Iterating a 1d array is similar to list and for loop to return rows. For element wise iteration use A.flat

```python

for row in E:
  print(row)
for element in E.flat:
  print(element)
```

For applying aggregrate function for columns or rows use apply_along_axis() function when axis=0 it operates on column and when axis=1 it operates on row.

```python
np.apply_along_axis(np.mean,axis=0,arr=E)
```

```python
np.apply_along_axis(np.mean,axis=1,arr=E)
```

```python
def half(x):
  return x/2
np.apply_along_axis(half,axis=0,arr=E)
```

Numpy uses the horizontal stacking hstack() to stack new array as columns while the vertical stacking vstack as rows.

```python
A1=np.ones((3,3))
A2=np.zeros((3,3))
np.vstack((A1,A2)) # You can also use concatenate function
```

```python
np.hstack((A1,A2)) # use concatenate((A1,A2),1)
```

column_stack and row_stack are used with 1d arrays stacking them to 2d array.

```python
d1=np.arange(0,3)
d2=np.arange(3,6)
d3=np.arange(6,9)
print(np.row_stack((d1,d2,d3)))
print(np.column_stack((d1,d2,d3)))
```

We can also split the array with vsplit and hsplit.

```python
I=np.arange(16).reshape(4,4)
[J,K]=np.vsplit(I,2)
print(J)
print(K)
```

```python
[L,M]=np.hsplit(I,2)
print(L)
print(M)
```

```python
# You can also split into nonsymmetric parts using axis where axis=0 it will be row operation and axis=1 it will be in column operation.
[I1,I2,I3]=np.split(I,[1,3],axis=1)
print(I1)
print(I2)
print(I3)
[I4,I5,I6]=np.split(I,[1,3],axis=0)
print(I4)
print(I5)
print(I6)
```
