Series can be thought as a Dictionary with key as index. Arithmetic operations can be done only for the common indices or else it will return NaN

```python
import pandas as pd
import numpy as np
s=pd.Series([1,2,2,1,0,3],['a','b','b','a','c','d'])
print(s)
sr=s.unique()
print(sr)
print(s.value_counts())
```
